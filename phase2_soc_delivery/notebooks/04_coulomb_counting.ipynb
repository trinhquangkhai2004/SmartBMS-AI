```python
from pathlib import Path
import os
import sys

root = Path.cwd().resolve()
if root.name == 'notebooks':
    root = root.parent
sys.path.insert(0, str(root))
from run_cc import run_baseline
archive = Path(os.environ.get('P22_ZIP', Path.home() / 'Downloads' / 'phase2_soc_P2_2_verified_delivery.zip'))
evidence = run_baseline(archive)
print(evidence['milestone'], evidence['gate'], 'ER4', evidence['ER4'])
```

```python
print('Run | Population | rows | MAE | RMSE | P95 | MAX (pp)')
for run_id, result in evidence['runs'].items():
    for population in ('FULL_DYNAMIC', 'NON_FLAGGED'):
        values = result[population]
        scores = [values[k] for k in ('MAE_pp', 'RMSE_pp', 'P95_abs_pp', 'MAX_abs_pp')]
        print(run_id, population, f"{values['numerator']}/{values['denominator']}",
              *('N/A' if score is None else f'{score:.4f}' for score in scores), sep=' | ')
```
